# PERFILADO DEL PRIMER DATASET

```python
# Importación del primer dataset
Data_Frame1 = pd.read_csv(r'/content/drive/MyDrive/UAIN1008R_Proyecto/datos/raw/Red vial nacional del Sistema Nacional de Carreteras 2022 a 2024 - Ministerio de Transportes y Comunicaciones – MTC.csv', encoding = 'latin1', sep=';')
print("Tamaño de la tabla: ", Data_Frame1.shape)

Data_Frame1.head()
```

```python
# Tipos de datos
Data_Frame1.dtypes
```

```python
# Total de nulos y % de nulos por columna
nulos = Data_Frame1.isnull().sum()
porcentaje_nulos = (Data_Frame1.isnull().mean() * 100).round(2)

tabla_nulos = pd.DataFrame({'Total Nulos': nulos, 'Porcentaje Nulos': porcentaje_nulos})

print("REPORTE DE VALORES NULOS: ")
tabla_nulos
```

```python
# Número de filas duplicadas
duplicados = Data_Frame1.duplicated().sum()

print("Número de filas duplicadas: ", duplicados)
```

```python
# Rangos inconsistentes
tabla_rangos1 = pd.DataFrame({
    'CATEGORIA': ['ID_RVN','ID_DEPARTAMENTO','INICIO','FINAL','LONGITUD','NRO_CARRILES','SUPERFICIE','ESTADO','CODIGO_CONCESION','FECHA_CORTE','ES_CONCES'],
    'MINIMO': [Data_Frame1['ID_RVN'].min(), Data_Frame1['ID_DEPARTAMENTO'].min(), Data_Frame1['INICIO'].min(), Data_Frame1['FINAL'].min(), Data_Frame1['LONGITUD'].min(), Data_Frame1['NRO_CARRILES'].min(), Data_Frame1['SUPERFICIE'].min(), Data_Frame1['ESTADO'].min(), Data_Frame1['CODIGO_CONCESION'].min(), Data_Frame1['FECHA_CORTE'].min(), Data_Frame1['ES_CONCES'].min()],
    'MAXIMO': [Data_Frame1['ID_RVN'].max(), Data_Frame1['ID_DEPARTAMENTO'].max(), Data_Frame1['INICIO'].max(), Data_Frame1['FINAL'].max(), Data_Frame1['LONGITUD'].max(), Data_Frame1['NRO_CARRILES'].max(), Data_Frame1['SUPERFICIE'].max(), Data_Frame1['ESTADO'].max(), Data_Frame1['CODIGO_CONCESION'].max(), Data_Frame1['FECHA_CORTE'].max(), Data_Frame1['ES_CONCES'].max()]
})

tabla_rangos1
```

```python
# Estadísticas descriptivas para columnas numéricas
Data_Frame1.describe()
```

```python
# Detección de outliers utilizando el método del IQR
numericos1 = Data_Frame1.select_dtypes(include='number')

q1_1 = numericos1.quantile(0.25)
q3_1 = numericos1.quantile(0.75)
iqr_1 = q3_1 - q1_1
outliers_1 = ((numericos1 < (q1_1 - 1.5 * iqr_1)) | (numericos1 > (q3_1 + 1.5 * iqr_1))).sum()

print("Número de outliers por columna:")
outliers_1
```

```python
# Función para ver los outliers por fila
mascara = (numericos1 < (q1_1 - 1.5 * iqr_1)) | (numericos1 > (q3_1 + 1.5 * iqr_1))
filas_outliers = Data_Frame1[mascara.any(axis=1)]
print(filas_outliers)
```

# Análisis de outliers:
Se aplicó el método del Rango Intercuartílico (IQR) para detectar valores atípicos en las columnas numéricas. Si bien se identificaron valores fuera del rango intercuartílico en columnas como LONGITUD, INICIO y FINAL, tras una revisión manual se determinó que estos corresponden a variaciones propias de la red vial nacional, donde coexisten tramos cortos de pocos metros junto a tramos de cientos de kilómetros según la región y jerarquía de la vía. Ningún valor resultó inválido, por lo que se decidió conservar todos los registros sin eliminación ni imputación.

```python
for col in Data_Frame1.select_dtypes(include='object').columns:
    print(f"\n{'='*40}")
    print(f"COLUMNA: {col}")
    print(f"Valores únicos: {Data_Frame1[col].nunique()}")
    vc = Data_Frame1[col].value_counts()
    raras = vc[vc < (len(Data_Frame1) * 0.01)]  # menos del 1%
    print(f"Categorías raras (< 1%): {len(raras)}")
    if len(raras) > 0:
        print(raras)
    else:
        print("Sin categorías raras ✅")
```

```python
# Tabla de métricas del primer Dataset

# -- Completitud: Porcentaje de valores no nulos
completitud1 = round((1 - Data_Frame1.isnull().sum().sum() / Data_Frame1.size) * 100, 2)

# -- Unicidad: Porcentaje de filas únicas
unicidad1 = round((1 - Data_Frame1.duplicated().sum() / len(Data_Frame1)) * 100, 2)

# -- Consistencia: Porcentaje de columnas sin problemas de rango o formato
cols_totales1 = 21
cols_con_problema1 = 3
consistencia1 = round(((cols_totales1 - cols_con_problema1) / cols_totales1) * 100, 2)

# Creación de la tabla de métricas
tabla_metricas1 = pd.DataFrame({
    'Dataset': ["Red vial nacional del Sistema Nacional de Carreteras 2022 a 2024"],
    'Filas': [len(Data_Frame1)],
    'Completitud': [completitud1],
    'Unicidad': [unicidad1],
    'Consistencia': [consistencia1]
})

print("Métricas de calidad del dataset:")
tabla_metricas1
```
